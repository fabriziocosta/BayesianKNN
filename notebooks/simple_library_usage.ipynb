# Simple library usage

Generate a dataset, split it, fit Bayesian model averaging with the default family registry, and evaluate the test set.

```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split

from bayesian_model_averaging import BayesianModelAveragingClassifier

X, y = make_moons(n_samples=500, noise=0.2, random_state=7)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=7, stratify=y
)

# The default family registry is used automatically.
# n_estimators is set lower here to keep the example quick.
model = BayesianModelAveragingClassifier(n_estimators=40, random_state=7)
model.fit(X_train, y_train)

print(f"test accuracy: {model.score(X_test, y_test):.3f}")
print("family posterior shares:", model.get_model_masses()["family"])
```
