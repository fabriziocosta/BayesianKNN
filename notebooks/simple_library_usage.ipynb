# Simple library usage

Load the wine dataset, split it, fit Bayesian model averaging with the default family registry, and evaluate the test set.

```python
import sys
from pathlib import Path

project_root = Path.cwd()
if not (project_root / "bayesian_model_averaging").is_dir() and (project_root.parent / "bayesian_model_averaging").is_dir():
    project_root = project_root.parent
if (project_root / "bayesian_model_averaging").is_dir():
    sys.path.insert(0, str(project_root))
```

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

from bayesian_model_averaging import BayesianModelAveragingClassifier

wine = load_wine()
X, y = wine.data, wine.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=7, stratify=y
)

# The default family registry is used automatically.
# n_estimators is set lower here to keep the example quick.
model = BayesianModelAveragingClassifier(n_estimators=100, random_state=7)
model.fit(X_train, y_train)

print(f"test accuracy: {model.score(X_test, y_test):.3f}")
family_shares = ", ".join(
    f"{name}={share:.3f}"
    for name, share in model.get_model_masses()["family"].items()
)
print(f"family posterior shares: {family_shares}")
```
